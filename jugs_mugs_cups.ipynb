# Type 1

```python
numbers = int(input("Enter a number: "))
print("Number:", numbers)

if numbers % 18 == 0:
    print("Jugs | Mugs | Cups")
elif numbers % 6 == 0:
    print("Jugs | Mugs")
elif numbers % 9 == 0:
    print("Jugs | Cups")
elif numbers % 3 == 0:
    print("Jugs")
else:
    print("No Jugs | Mugs | Cups")
```

# Type 2

```python
val = int(input("Enter the number:"))
if val % 9 == 0:
    print("cups")
elif val % 6 == 0 and val % 9 != 0:
    print("mugs")
elif val % 3 == 0 and (val % 6 != 0 or val % 9 != 0):
    print("jugs")
else:
    print(val, "Not matched")
```

# Type 3

```python
data = int(input("Enter the number:"))
temp = str(data)
store = []
#print(type(temp))
for i in range(len(temp)):
    if temp[i] == '3':
        store.append('Jugs')
        #print("Jugs")
    elif temp[i] == '6':
        store.append('Mugs')
        #print("Mugs")
    elif temp[i] == '9':
        store.append('Cups')
        #print("Cups")
    else:
        store.append(temp[i])
        #print(i)
print(store)
```

# Type 4

```python
new_data = int(input("Enter the number:"))
temp_data = str(new_data)
store_data = []
for j in range(len(temp_data)):
    if int(temp_data[j]) % 9 == 0 and int(temp_data[j]) % 6 == 0:
        store_data.append('Mugs | Cups')
    elif int(temp_data[j]) % 6 == 0 and int(temp_data[j]) % 3 == 0:
        store_data.append('Jugs | Mugs')
    elif int(temp_data[j]) % 9 == 0 and int(temp_data[j]) % 3 == 0:
        store_data.append('Jugs | Cups')
    elif int(temp_data[j]) % 3 == 0:
        store_data.append('Jugs')
    else:
        store_data.append(temp_data[j])
print(store_data)
```
