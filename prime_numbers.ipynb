# Question 1
> Write a Python program to check whether a given number is a prime number or not.

**Hint**
* A prime number has exactly two factors.
* Check divisibility from 2 to n-1.
* If divisible even once, it is not prime

> Sample Input
> **Enter a number: 13**
> Sample Output
> **13 is a Prime Number**

# Q1.Solution

```python
def prime(n):
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False        
    for i in range(5, int(n**0.5) + 1, 6):
        if n % i == 0 or n % (i + 2) == 0:
            return False            
    return True
val = int(input())
prime(val)
if val == True:
    print(val,"is a Prime Number")
print(val,"is a Not Prime Number")
```

# Question 2
> Write a program to print all prime numbers between 1 and N.

**Hint**
* Use one loop to go from 2 to N.
* For each number, check whether it is prime.

> Sample Input **Enter N: 20** Sample Output **2 3 5 7 11 13 17 19**

# Q2.Solution

```python
n = int(input("Enter N: "))

for num in range(2, n + 1):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```
